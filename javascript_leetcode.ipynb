# 1. Retornar o índice da soma de dois números de uma lista. 

## 1.1 Força bruta:
   A complexidade desse algorito é de $O(n²)$.

```javascript
/**
 * @param {number[]} nums
 * @param {number} target
 * @return {number[]}
 */
var twoSum = function(nums, target) {
    for (var i=0; i<nums.length; i++){
        for (var j=i+1; j<nums.length; j++){
            if (nums[i] + nums[j] == target){
                return [i,j];
            } 
        }
    }
};
```
